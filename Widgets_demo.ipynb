```python
#Creating a Dropdown in widgets for listing and displaying ADLS files based on folder
path = "abfss://kanak@adlsdatawc.dfs.core.windows.net/data/"

items = [f.name.rstrip('/') for f in dbutils.fs.ls(path)]

dbutils.widgets.dropdown(
    "selected_folder",
    items[0],   # default value
    items,
    "Select Folder"
)

selected_folder = dbutils.widgets.get("selected_folder")
print(selected_folder)

file_path = f"{path}{selected_folder}"

# Read and display
df = spark.read.option("header", "true").csv(file_path)

display(df.limit(5))
```
